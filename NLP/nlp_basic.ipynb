```python
chat_word_dict={
    "brb": "be right back",
    "btw": "by the way",
    "idk": "I don't know",
    "imo": "in my opinion",
    "lmao": "laughing",
    "lol": "laughing out loud",
    "omg": "oh my god",
    "tbh": "to be honest",
    "ttyl": "talk to you later",
    "u": "you",
    "ur": "your",
    "pls": "please",
    "thx": "thanks",
    "wbu": "what about you",
    "xoxo": "hugs and kisses",
    "smh": "shaking my head",
    "fyi": "for your information",
    "bff": "best friends forever",
    "fomo": "fear of missing out",
    "tmi": "too much information",
    "lmk": "let me know",
    "wyd": "what are you doing",
    "np": "no problem",
    "glhf": "good luck have fun",
    "sry": "sorry",
    "gm": "good morning",
    "gn": "good night",
    "cya": "see you",
    "jk": "just kidding",
    "b4": "before",
    "gr8": "great",
    "bday": "birthday",
    "imo": "in my opinion",
    "l8r": "later",
    "plz": "please",
    "thx": "thanks",
    "w8": "wait",
    "wtf": "what the fuck",
    "ily": "I love you",
    "ttys": "talk to you soon",
    "hbu": "how about you",
    "bbl": "be back later",
    "gtg": "got to go",
    "ppl": "people",
    "wanna": "want to",
    "gonna": "going to",
    "lemme": "let me",
    "dunno": "don’t know"
}
```

```python
chat_text = "brb, I will ttyl. btw, idk if u r coming to the party."
def normalize_chat_text(chat_text):
    words = chat_text.split()
    normalized_words = [chat_word_dict.get(word.lower(), word) for word in words]      #when we assign key to get functions then it gives its value, if not then it returns the word itself
    return ' '.join(normalized_words)
```

```python
normalized_text = normalize_chat_text(chat_text="brb, I will ttyl. btw, idk if u r coming to the party.")
```
